```python
# =========================
# IMPORT LIBRARY
# =========================
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# =========================
# UPLOAD GAMBAR
# =========================
uploaded = files.upload()

# Ambil nama file
file_name = list(uploaded.keys())[0]

# =========================
# BACA CITRA GRAYSCALE
# =========================
img = cv2.imread(file_name, cv2.IMREAD_GRAYSCALE)

# =========================
# TRANSFORMASI NEGATIF
# =========================
negatif = 255 - img

# LOG TRANSFORM FINAL (REKOMENDASI DOSEN LEVEL)
img_float = img.astype(np.float32)

log_img = np.log1p(img_float)

# scaling + contrast adjustment
log_img = (log_img / np.max(log_img)) * 255

# OPTIONAL: biar lebih kontras
log_img = cv2.equalizeHist(log_img.astype(np.uint8))

# =========================
# TAMPILKAN GAMBAR
# =========================
plt.figure(figsize=(10,8))

plt.subplot(2,2,1)
plt.imshow(img, cmap='gray')
plt.title('Citra Asli')
plt.axis('off')

plt.subplot(2,2,2)
plt.imshow(negatif, cmap='gray')
plt.title('Negatif')
plt.axis('off')

plt.subplot(2,2,3)
plt.imshow(log_img, cmap='gray')
plt.title('Logaritmik')
plt.axis('off')

# =========================
# HISTOGRAM
# =========================
plt.subplot(2,2,4)
plt.hist(img.ravel(), bins=256, range=[0,256])
plt.title('Histogram Asli')

plt.tight_layout()
plt.show()
```

**ANALISA SOAL NO 1**

Citra asli memiliki distribusi intensitas grayscale yang normal dengan dominasi pada nilai menengah.

Pada transformasi negatif, nilai intensitas dibalik sehingga area terang menjadi gelap dan area gelap menjadi terang. Hal ini terlihat dari perubahan kontras pada citra yang berlawanan dengan citra asli.

Pada transformasi logaritmik, intensitas rendah (area gelap) diperbesar sehingga detail pada bagian gelap menjadi lebih terlihat. Namun tanpa normalisasi, citra dapat menjadi terlalu terang. Setelah dilakukan normalisasi, distribusi intensitas menjadi lebih seimbang dan kontras tetap terjaga.

Histogram menunjukkan persebaran intensitas piksel. Pada citra asli, distribusi cenderung berada di nilai menengah, sedangkan pada transformasi logaritmik distribusi bergeser sehingga area gelap lebih terangkat.

Kesimpulannya, transformasi negatif cocok untuk analisis kontras, sedangkan transformasi logaritmik lebih efektif untuk meningkatkan detail pada area gelap.
