```python
# =========================
# IMPORT LIBRARY
# =========================
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# =========================
# UPLOAD GAMBAR
# =========================
uploaded = files.upload()
file_name = list(uploaded.keys())[0]

# =========================
# BACA CITRA
# =========================
img = cv2.imread(file_name, cv2.IMREAD_GRAYSCALE)

# =========================
# DEFINISI KERNEL
# =========================

# LPF
lpf = (1/9) * np.ones((3,3))

# HPF
hpf = np.array([
    [-1,-1,-1],
    [-1, 8,-1],
    [-1,-1,-1]
])

# BPF (Sharpen)
bpf = np.array([
    [0,-1,0],
    [-1,5,-1],
    [0,-1,0]
])

# =========================
# KONVOLUSI
# =========================
img_lpf = cv2.filter2D(img, -1, lpf)
img_hpf = cv2.filter2D(img, -1, hpf)
img_bpf = cv2.filter2D(img, -1, bpf)

# =========================
# TAMPILKAN HASIL
# =========================
plt.figure(figsize=(10,8))

plt.subplot(2,2,1)
plt.imshow(img, cmap='gray')
plt.title('Original')
plt.axis('off')

plt.subplot(2,2,2)
plt.imshow(img_lpf, cmap='gray')
plt.title('LPF (Blur)')
plt.axis('off')

plt.subplot(2,2,3)
plt.imshow(img_hpf, cmap='gray')
plt.title('HPF (Edge)')
plt.axis('off')

plt.subplot(2,2,4)
plt.imshow(img_bpf, cmap='gray')
plt.title('BPF (Sharpen)')
plt.axis('off')

plt.tight_layout()
plt.show()
```

**ANALISA SOAL NO 2**

Low Pass Filter (LPF) menghasilkan citra yang lebih halus karena mengurangi noise dengan cara merata-ratakan nilai piksel. Namun detail halus pada citra juga ikut berkurang.

High Pass Filter (HPF) menonjolkan tepi dengan memperkuat perubahan intensitas antar piksel. Hasilnya citra terlihat hanya bagian edge atau garis tepi.

Band Pass Filter (BPF) atau sharpening meningkatkan ketajaman citra dengan mempertahankan detail sekaligus memperkuat tepi.

Dari sisi nilai piksel, LPF cenderung mempertahankan rata-rata, HPF menghasilkan nilai ekstrem (tinggi/rendah), sedangkan BPF meningkatkan kontras lokal.

Dalam domain frekuensi, LPF melewatkan frekuensi rendah, HPF melewatkan frekuensi tinggi, dan BPF merupakan kombinasi keduanya.
